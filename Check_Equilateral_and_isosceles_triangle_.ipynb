#The triangle is equilateral

```python
a=float(input("Enter a:"))
b=float(input("Enter b:"))
c=float(input("Enter c:"))
if a==b==c:
  print("The triangel is equilateral")
else:
  print("The triangel is not equilateral")
```

# The triangel's is isoscale

```python
a=float(input("Enter a:"))
b=float(input("Enter b:"))
c=float(input("Enter c:"))
if a==b or b==c or c==a:
  print("The triangel's is isoscale")
else:
  print("The triangel's is not isoscale")
```
